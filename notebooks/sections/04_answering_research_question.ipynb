<h1>4. Answering of Research Question</h1>

<p>To revisit and for quick reference purpose, I will be indicating the Research Question of the analysis.</p>
<br/>
<p style="text-align:center;"><strong>What factors make job postings more visible and increase application rates, and how can job seekers use these insights to improve their job search?</strong></p>
<br/>
<p>I will be answering the Research Question in point form so as to address different insights obtained.</p>
<br/>
<ul>
    <li>Job Posting typically last for around 1 month as that is normally when Companies / Employers have sufficient applicants and it is also most likely a common industry practice. Therefore Job-seekers should actively (min. Every 2 weeks) go to job portals such as LinkedIn to find new Job Postings that they might be interested in and apply as soon as possible especially if the Job Post is close to 1 Month old.</li>
    <br/>
    <li>Many Companies / Employers are not providing remote work which is the preferance after the pandemic. This can lead to more applicants for Remote Work Job Postings and lesser applicants for Job Postings that do not offer Remote Work. Job-seekers can cease this opportunity to apply for Jobs that do not offer Remote Work especially if Job arrangement is not a major concern. However, if a Job-seeker wish to apply for the limited amount of Job Posting that offer Remote Work, they should apply as soon as possible with a impressive resume to secure the job.</li>
    <br/>
    <li>Companies / Employers should take note that despite more views normally reflecting more applicants in most context. Most viewers are actually not applicants which means Companies / Employers should continue to actively seek out even more viewers as it may not reflect their actual applicant pool. Job-seekers can cease this ratio difference as an opportunity to apply for popular jobs with large amount of views as the popularity may not neccessarily reflect the total application pool.</li>
    <br/>
    <li>Job Postings with Lower and Higher Salary Range have higher Apply-to-View Ratio as Job-seekers value Salary Transparency and Salary Negotiation. Should a Job-seeker feel that Salary Transparency and / or Salary Negotiation may not be their key priority, they can apply for a job with medium salary range which have lower applicants and also potentially a balance between Salary Transparency and Salary Negotiation. However, if a Job-seeker wish to have Salary Transparency and / or Salary Negotiation, they will need to apply as soon as possible and prepare a more outstanding resume due to the large quantity of applicants.</li>
    <br/>
    <li>Job Posting with Lower Experience Level has higher Apply-to-View Ratio most likely due to the lower qualifications required. Despite Higher Experience Level Job Postings having a higher salary, they have much fewer applicants compared to the other Experience Level. This means that Job-seekers who are applying for Low to Medium Experience Level Jobs should know that they have a larger competitor pool with lower qualifications requirement whereas Job-seekers who are applying for High Experience Level Jobs have a smaller competitor pool with higher qualifications requirement.</li>
    <br/>
    <li>Most Job Postions does not have Remote Working arrangements for High Experience Level Jobs and Internship. Employers can trial a few positions within these Experience Level to evaluate the productivity which can lead to more applicants for these positions. Job-seekers who intend to work Remotely should try to apply to Mid-Senior Level Positions as there are more Remote Working arrangement Job Postings there.</li>
    <br/>
    <li>Job Posting offering Remote Work arrangements have more then 3 times as much applicants as Job Posting which does not offer Remote Work arrangements. This means that Employers if possible should allow for Remote Work as it attract more applicants. As for Job-seekers, they shoul be aware that should they be applying for Jobs which allow for Remote Work arrangements, they will be facing a larger pool of applicants as compared to On-site Working arrangement. Thus, if On-site working is not an issue, Job-seekers should try to apply for Job Posting with On-site Working arrangements to secure a Job much more easily.</li>
    <br/>
    <li>Job Posting which are up longer typically have a larger apply-to-view ratio which peaks at 4 Months. Therefore, despite industry practices, Companies / Employers should put out Job Postings for up to 4 Months if they require a large pool of applicants. Job-seekers should try to apply as soon as possible especially if a large number of applicants is expected, so as to stand out from all the other applicants as an 'early applicants'.</li>
</ul>
