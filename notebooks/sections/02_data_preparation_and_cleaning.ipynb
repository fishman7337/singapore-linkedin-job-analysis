<h3>2.4.1 Identifying Missing Data and Imputation Strategy</h3>

<p>Before imputing the data, I will identify the data columns with missing data one last time and subsequently, decide which imputation strategy will be most suitable in regards to the data columns.</p>

```python
# ----------- Renaming of Data Columns ----------- #
merged_df = merged_df.rename(columns={'company_name_x': 'company_name', 'title_x': 'title'})

# ----------- Placing Missing Values of the Merged Dataframe Into A Dataframe ----------- #
mergedDfMissingData = pd.DataFrame({'Missing Values': merged_df.isnull().sum().values}, index = merged_df.columns)

# ----------- Function for Colour Coding ----------- #
def colour_code(val):
    if val > 0:
        return 'background-color: red; color: white'
    else:
        return 'background-color: green; color: white'

# ----------- Calling Function for Colour Coding ----------- #
mergedDfMissingData = mergedDfMissingData.style.map(colour_code, subset = ['Missing Values'])

# ----------- Displaying Missing Values----------- #
mergedDfMissingData
```

<p>With reference to the output above, I have made an action report on how I shall proceed with the imputation. The action report is listed below.</p>
<br/>
<p><strong><u>company_name</u></strong></p>
<ul>
    <li>Imputation Strategy: String Filling</li>
    <li>Reason: Company name cannot be imputed. Therefore, a placeholder will be used.</li>
</ul>
<p><strong><u>max_salary</u></strong></p>
<ul>
    <li>Imputation Strategy: Group Interpolation with Experience Level and Pay Period</li>
    <li>Reason: Due to potential outliers and large number of Missing Data and requiring a realistic value.</li>
</ul>
<p><strong><u>pay_period</u></strong></p>
<ul>
    <li>Imputation Strategy: Mode Imputation</li>
    <li>Reason: Mode will ensure that the most frequent value will be used.</li>
</ul>
<p><strong><u>views</u></strong></p>
<ul>

<h3>2.4.2 Imputation of Missing Data</h3>

<p>Using the identified strategy listed in the previous section, I will be using the identified strategy on the data column with missing data.</p>

```python
# ----------- String Filling ----------- #
merged_df['company_name'] = merged_df['company_name'].fillna('Unknown')

# ----------- Integer Filling ----------- #
merged_df['views'] = merged_df['views'].fillna(0)
merged_df['applies'] = merged_df['applies'].fillna(0)
merged_df['remote_allowed'] = merged_df['remote_allowed'].fillna(0)

# ----------- Mode Imputation ----------- #
merged_df['pay_period'] = merged_df['pay_period'].fillna(merged_df['pay_period'].mode()[0])
merged_df['formatted_experience_level'] = merged_df['formatted_experience_level'].fillna(
    merged_df['formatted_experience_level'].mode()[0]
)

# ----------- Function for Interpolation by Groups ----------- #
def interpolate_salary_by_group(df, column):
    # ----------- Sorting of Data ----------- #
    df = df.sort_values(by = ['pay_period', 'formatted_experience_level', 'listed_time'])
    
    # ----------- Interpolation Process ----------- #
    df[column] = df.groupby(['pay_period', 'formatted_experience_level'])[column] \
                   .transform(lambda x: x.interpolate(method = 'linear', limit_direction = 'both'))
    
    # ----------- Returning of Final Results ----------- #
    return df

# ----------- Interpolation for max_salary and min_salary ----------- #
merged_df = interpolate_salary_by_group(merged_df, 'max_salary')
merged_df = interpolate_salary_by_group(merged_df, 'min_salary')

# ----------- Placing Missing Values of the Merged Dataframe Into A Dataframe ----------- #
mergedDfMissingData = pd.DataFrame({'Missing Values': merged_df.isnull().sum().values}, index = merged_df.columns)

# ----------- Function for Colour Coding ----------- #
def colour_code(val):
    if val > 0:
        return 'background-color: red; color: white'
    else:
        return 'background-color: green; color: white'

# ----------- Calling Function for Colour Coding ----------- #
mergedDfMissingData = mergedDfMissingData.style.map(colour_code, subset = ['Missing Values'])

# ----------- Displaying Missing Values----------- #
mergedDfMissingData
```

$$
\text{max\_salary\_normalised} = \text{max\_salary (Monthly Pay)} \times 12
$$
<br/>
$$
\text{max\_salary\_normalised} = \text{max\_salary (Biweekly Pay)} \times 26

$$
\text{max\_salary\_normalised} = \text{max\_salary (Weekly Pay)} \times 52
$$

$$
\text{max\_salary\_normalised} = \text{max\_salary (Hourly Pay)} \times 40 \times 52
$$

$$
\text{min\_salary\_normalised} = \text{min\_salary (Monthly Pay)} \times 12
$$
<br/>
$$
\text{min\_salary\_normalised} = \text{min\_salary (Biweekly Pay)} \times 26
$$

$$
\text{min\_salary\_normalised} = \text{min\_salary (Weekly Pay)} \times 52

```python
# ----------- Function to Normalise Salary With Respect to Pay Period ----------- #
def normalise_salary_max(row):
    if row['pay_period'] == 'Hourly':
        return row['max_salary'] * 40 * 52  # 40 Hours/Week, 52 Weeks/Year
    elif row['pay_period'] == 'Monthly':
        return row['max_salary'] * 12       # 12 Months/Year
    elif row['pay_period'] == 'Biweekly':
        return row['max_salary'] * 26       # 26 Weeks/Year
    elif row['pay_period'] == 'Weekly':
        return row['max_salary'] * 52       # 52 Weeks/Year
    else:
        return row['max_salary']            # No Conversion Needed

# ----------- Creation of New Data Column 'max_salary_normalised' ----------- #
merged_df['max_salary_normalised'] = merged_df.apply(normalise_salary_max, axis=1)

# ----------- Function to Normalise Salary With Respect to Pay Period ----------- #
def normalise_salary_min(row):
    if row['pay_period'] == 'Hourly':
        return row['min_salary'] * 40 * 52  # 40 hours/week, 52 weeks/year
    elif row['pay_period'] == 'Monthly':
        return row['min_salary'] * 12       # 12 months/year
    elif row['pay_period'] == 'Biweekly':
        return row['min_salary'] * 26       # 26 Weeks/Year
    elif row['pay_period'] == 'Weekly':
        return row['min_salary'] * 52       # 52 Weeks/Year
    else:
        return row['min_salary']            # No conversion needed

# ----------- Creation of New Data Column 'min_salary_normalised' ----------- #
merged_df['min_salary_normalised'] = merged_df.apply(normalise_salary_min, axis = 1)

# ----------- Posting Duration as Alias postingDuration ----------- #
merged_df['postingDuration'] = (merged_df['expiry'] - merged_df['listed_time']).dt.days

# ----------- Salary Range as Alias salaryRange ----------- #
merged_df['salary_range'] = merged_df['max_salary_normalised'] - merged_df['min_salary_normalised']

# ----------- Apply to View Ratio as Alias applyViewRation ----------- #
merged_df['apply_view_ratio'] = merged_df['applies'] / merged_df['views']
merged_df['apply_view_ratio'] = merged_df['applies'] / merged_df['views'].replace(0, np.nan)
merged_df['apply_view_ratio'] = merged_df['apply_view_ratio'].fillna(0)  # Replace NaN with 0 after calculation


# ----------- Displaying Dataframe ----------- #
merged_df.head(5)
```

<h3>2.9.1 Handling Outliers Using IQR Method</h3>

<p>The IQR Method requires the establishment of a Upper Fence and Lower Fence such that any data which are above the Upper Fence will be deemed invalid and illegitimate and will be removed whereas any data which are below the Lower Fence will be deemed likewise and removed. In the cells below, I will be removing Outliers using the IQR Method and subsequently comparing the effectiveness and results with the Z-Score Method. The formula for establishing the Upper Fence, Lower Fence and IQR are listed below.</p>

$$\text{IQR} = \text{Q3} - \text{Q1}$$
$$\text{Lower Fence} = \text{Q1} - 1.5 \times \text{IQR}$$
$$\text{Upper Fence} = \text{Q3} + 1.5 \times \text{IQR}$$
<br/>
<p>With the formulas listed, I will proceed with identifying and removing the Outliers.</p>

<h4>2.9.1.4 Removing Identified Outliers Using IQR Method</h4>

<p>After identifying large amounts of Outliers, I will now proceed to remove the Outliers from the dataframe. However, I will be doing this operation on a copy of the original dataframe to ensure I am still able to do the Outliers Identification and Removal using the Z-Score method.</p>

```python
# ----------- Copy the Main DataFrame ----------- #
IQRdf = merged_df.copy()

# ----------- Define a Function to Remove Outliers Using IQR ----------- #
def remove_outliers_iqr(df, column):
    # Calculate Q1, Q3, and IQR
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    # Define lower and upper bounds
    lowerFence = Q1 - 1.5 * IQR
    upperFence = Q3 + 1.5 * IQR
    
    # Filter DataFrame to keep only non-outliers
    df_no_outliers = df[(df[column] >= lowerFence) & (df[column] <= upperFence)]
    
    print(f"Outliers removed for '{column}': {len(df) - len(df_no_outliers)} rows")
    return df_no_outliers

# ----------- Remove Outliers for Each Column ----------- #
IQRdf = remove_outliers_iqr(IQRdf, 'min_salary_normalised')
IQRdf = remove_outliers_iqr(IQRdf, 'max_salary_normalised')
IQRdf = remove_outliers_iqr(IQRdf, 'salary_range')

# ----------- Display Results ----------- #
print(f"Rows after outlier removal: {len(IQRdf)}")
IQRdf.head(5)
```

<h3>2.9.2 Handling Outliers Using Z-Score</h3>

<p>In the previous section, I have used the IQR method to identify and address Outliers. However, as there are many methods to identify and address Outliers, I will be using the Z-Score method in this section. After removing Outliers using the Z-score Method, I will then cross refer the result and verify which method is most suitable. The formula for Z-Score is indicated below.</p>

$$
Z = \frac{X - \mu}{\sigma}
$$

\begin{align*}
Z & : \text{Z-Score (standardised value)} \\
X & : \text{Observed data point} \\
\mu & : \text{Mean of the data} \\
\sigma & : \text{Standard deviation of the data}
\end{align*}

<br>
<p>With the formula indicated, I will proceed to identify the Outliers and remove the Outliers in the following cells.</p>

<h4>3.9.2.4 Removing Identified Outliers Using Z-Score Method</h4>

<p>After identifying some Outliers within the dataframe using the Z-Score method, I will now proceed to remove them from the dataframe. However, I will be using a separate dataframe so as to allow for comparison afterwards.</p>

```python
# ----------- Copy DataFrame for Outlier Removal ----------- #
zScoreDf = merged_df.copy()

# ----------- Define Function to Remove Outliers Using Z-Score ----------- #
def remove_outliers_zscore(df, column, threshold = 3):
    mean = df[column].mean()
    std = df[column].std()
    
    # ----------- Calculating Z-Score ----------- #
    df['z_score'] = (df[column] - mean) / std
    
    # ----------- Removing Values Exceeding Threshold ----------- #
    df_no_outliers = df[df['z_score'].abs() <= threshold]
    
    print(f"Outliers removed for '{column}': {len(df) - len(df_no_outliers)} rows")
    return df_no_outliers.drop(columns=['z_score'])

# ----------- Apply Function for All Three Columns ----------- #
zScoreDf = remove_outliers_zscore(zScoreDf, 'min_salary_normalised')
zScoreDf = remove_outliers_zscore(zScoreDf, 'max_salary_normalised')
zScoreDf = remove_outliers_zscore(zScoreDf, 'salary_range')

# ----------- Display Results ----------- #
print(f"Rows after outlier removal: {len(zScoreDf)}")
zScoreDf.head(5)
```

<p>With reference to the output of the dataframe above, I am able to see that the removal of Outliers have been conducted successfully and I am able to perform comparisiom between the two Outliers Removal Operation.</p>
